# A fancy function game

```python
import numpy as np
```

## Level 1

```python
# Game items
input_items = np.array(["A", "B", "C", "D", "E"])
n = len(input_items)

# Game start
setup_indices = np.random.choice(n, n, replace=False)
setup_items = np.array(input_items)[setup_indices]

print("The game starts with a random permutation of some items:\n")
print("\t\t", setup_items)

layer_indices = np.random.choice(n, n, replace=False)
dummy_indices1 = dummy_indices2 = layer_indices.copy()

while np.all(dummy_indices1 == layer_indices) or np.all(dummy_indices2 == layer_indices) or np.all(dummy_indices1 == dummy_indices2):
    dummy_indices1 = np.random.choice(n, n, replace=False)
    dummy_indices2 = np.random.choice(n, n, replace=False)

middle_layer = np.array([layer_indices, dummy_indices1, dummy_indices2])[np.random.choice(3, 3, replace=False)]

print("\nWe are given three arrays of indices in the middle layer:\n")
print("\t", middle_layer[0] + 1, "\t", middle_layer[1] + 1, "\t", middle_layer[2] + 1)

output_items = setup_items[layer_indices]

print("\nThe output layer shows the inventory permuted:\n")
print("\t\t", output_items)

print("\nWhich permutation is the correct one?", "[l] for left, [c] for centre, [r] for right.")
user_answer = input("[l] for left, [c] for centre, [r] for right.")

answer_key = {
    "l": middle_layer[0],
    "c": middle_layer[1],
    "r": middle_layer[2]
}

right_answer = [k for k, v in answer_key.items() if np.all(v == layer_indices)][0]

if user_answer == right_answer:
    print(f"\nTrue enough. {right_answer} is the answer.")
else:
    print(f"\nFalse. {right_answer} is the answer.")
```

## Level 2
